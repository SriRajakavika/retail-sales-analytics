```python
status_counts = orders["order_status"].value_counts()

fig, ax = plt.subplots(figsize=(7, 5))

wedges, texts, autotexts = ax.pie(
    status_counts.values,
    startangle=90,
    autopct="%1.1f%%",
    pctdistance=1.12,
    radius=0.75
)

for autotext in autotexts:
    autotext.set_fontsize(9)
    autotext.set_weight("bold")

ax.set_title("Order Status Distribution", fontsize=14)

ax.legend(
    wedges,
    status_counts.index,
    title="Order Status",
    loc="center left",
    bbox_to_anchor=(1, 0.5),
    fontsize=9
)

plt.tight_layout()
plt.show()
```

```python
state_counts = customers["state"].value_counts().head(10)

plt.figure(figsize=(8, 5))

plt.bar(state_counts.index, state_counts.values)

plt.title("Top 10 States by Customer Count")
plt.xlabel("State")
plt.ylabel("Number of Customers")

plt.xticks(rotation=0, ha="right")

plt.tight_layout()
plt.show()
```

```python
orders['order_date'] = pd.to_datetime(orders['order_date'],dayfirst=True)
```

```python
orders['order_date'].head()
```

```python
merged_df = pd.merge(orders, order_items, on='order_id')
```

```python
merged_df.head()
```

```python
reference_date = merged_df['order_date'].max() + pd.Timedelta(days=1)
reference_date
```

```python
rfm_data = merged_df.groupby('customer_id').agg({'order_date': lambda x: (reference_date - x.max()).days,
'order_id': 'nunique','total_price': 'sum'}).reset_index()
```

```python
rfm_data.columns = ['customer_id', 'Recency', 'Frequency', 'Monetary']
```

```python
rfm_data.head()
```

```python
rfm_data['Frequency'].value_counts()
```

```python
rfm_data['Segment'] = 'Regular'

rfm_data.loc[
    (rfm_data['Recency'] <= rfm_data['Recency'].quantile(0.25)) &
    (rfm_data['Frequency'] >= rfm_data['Frequency'].quantile(0.75)),
    'Segment'
] = 'Loyal'

rfm_data.loc[
    rfm_data['Recency'] >= rfm_data['Recency'].quantile(0.75),
    'Segment'
] = 'At Risk'
```

```python
rfm_data['Segment'].value_counts()
```

```python
rfm_data.to_sql(
    'customer_segments',
    con=engine,
    if_exists='replace',
    index=False)
print("customer_segments table exported successfully!")
```

```python
rfm_data.to_csv("customer_segments.csv", index=False)
```
